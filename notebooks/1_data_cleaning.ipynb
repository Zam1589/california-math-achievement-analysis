```python
import pandas as pd
```

```python
math = pd.read_excel("mathdownload2025.xlsx")
chronic = pd.read_excel("chronicdownload2025.xlsx")
```

```python
math_school = math[
    (math["rtype"] == "S") &
    (math["studentgroup"] == "ALL")
].copy()
```

```python
chronic_school = chronic[
    (chronic["rtype"] == "S") &
    (chronic["studentgroup"] == "ALL")
].copy()
```

```python
chronic_school.head()
```

```python
math_school.columns.tolist()
```

```python
math_school[[
    "schoolname",
    "currdenom",
    "currstatus"
]].head(10)
```

```python
merged = math_school.merge(
    chronic_school,
    on="cds",
    how="inner",
    suffixes=("_math", "_chronic")
)
```

```python
merged.shape
```

```python
merged[[
    "schoolname_math",
    "currstatus_math",
    "currstatus_chronic"
]].head(10)
```

```python
merged[[
    "currstatus_math",
    "currstatus_chronic"
]].isna().sum()
```

```python
merged[[
    "currstatus_math",
    "currstatus_chronic"
]].describe()
```

```python
analysis = merged.dropna(
    subset=["currstatus_math", "currstatus_chronic"]
).copy()
```

```python
analysis.shape
```

```python
analysis[[
    "schoolname_math",
    "currstatus_math",
    "currstatus_chronic"
]].head(10)
```

```python
analysis[[
    "currstatus_math",
    "currstatus_chronic"
]].isna().sum()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    analysis["currstatus_chronic"],
    analysis["currstatus_math"],
    alpha=0.3
)

plt.xlabel("Chronic Absenteeism Rate (%)")
plt.ylabel("Math Distance From Standard")
plt.title("Chronic Absenteeism and Mathematics Achievement\nCalifornia Public Schools, 2024–25")

plt.axhline(0, linestyle="--")

plt.show()
```

```python
analysis[["currstatus_chronic", "currstatus_math"]].corr()
```

```python
import numpy as np

x = analysis["currstatus_chronic"]
y = analysis["currstatus_math"]

slope, intercept = np.polyfit(x, y, 1)

plt.figure(figsize=(8, 6))

plt.scatter(x, y, alpha=0.25)

plt.plot(
    x,
    slope * x + intercept
)

plt.xlabel("Chronic Absenteeism Rate (%)")
plt.ylabel("Math Distance From Standard")
plt.title(
    "Chronic Absenteeism and Mathematics Achievement\n"
    "California Public Schools, 2024–25"
)

plt.axhline(0, linestyle="--")

plt.show()
```

```python
import statsmodels.api as sm

X = analysis["currstatus_chronic"]
Y = analysis["currstatus_math"]

X = sm.add_constant(X)

model = sm.OLS(Y, X).fit()

print(model.summary())
```
